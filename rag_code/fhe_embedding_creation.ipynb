```python
%load_ext autoreload
%autoreload 2

import os
from fastembed import TextEmbedding
from vectordb import VectorDB
import numpy as np
import json

# Sie können hier ihren eigenen HuggingFace Token einsetzen für schneller Downloads der Modelle (Man muss sich dafür nur Registrieren)
os.environ["HF_TOKEN"] = ""
```

```python
true_data = None
with open("unpoisoned_data.json", "r", encoding="utf-8") as f:
    true_data = json.load(f)

model = TextEmbedding(model_name="jinaai/jina-embeddings-v2-base-en")

client_questions = []
server_data = []
labels = []
for pair in true_data:
    question_embed = next(model.embed(pair["question"]))
    question_embed = VectorDB.normalize(question_embed).tolist()
    answer_embed = next(model.embed(pair["answer"]))
    answer_embed = VectorDB.normalize(answer_embed).tolist()
    client_questions.append({"question" : pair["question"], "embed" : question_embed})
    answer_embed = next(model.embed(pair["answer"]))
    answer_embed = VectorDB.normalize(answer_embed).tolist()
    server_data.append({"data" : pair["answer"], "embed" : answer_embed})
    labels.append({"question" : pair["question"], "answer" : pair["answer"]})
    
    
with open("client_questions.json", "w") as f:
    json.dump(client_questions, f)

with open("server_data.json", "w") as f:
    json.dump(server_data, f)

with open("labels.json", "w") as f:
    json.dump(labels, f)
```
